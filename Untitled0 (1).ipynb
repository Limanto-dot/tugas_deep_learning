```python
import json
import random
import time
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
from IPython.display import display
from sklearn.metrics import f1_score
from sklearn.model_selection import train_test_split
from torch.utils.data import ConcatDataset, DataLoader, Dataset
from torchvision import datasets, models, transforms

SEED = 42
BATCH_SIZE = 32
MAX_EPOCHS = 25
PATIENCE = 4
WEIGHT_DECAY = 1e-2
NUM_WORKERS = 0

PROJECT_DIR = Path.cwd()
DATA_DIR = PROJECT_DIR / "data"
from google.colab import drive

drive.mount("/content/drive")

RESULTS_DIR = Path(
    "/content/drive/MyDrive/Tugas05/results"
)
DATA_DIR.mkdir(parents=True, exist_ok=True)
RESULTS_DIR.mkdir(parents=True, exist_ok=True)


def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False


set_seed()
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
WEIGHTS = models.ResNet18_Weights.IMAGENET1K_V1
```

```python
trainval_raw = datasets.OxfordIIITPet(root=DATA_DIR, split="trainval", target_types="category", download=True)
test_raw = datasets.OxfordIIITPet(root=DATA_DIR, split="test", target_types="category", download=True)
assert trainval_raw.class_to_idx == test_raw.class_to_idx, "Pemetaan label dua split berbeda"

SELECTED_CLASSES = [
    "Abyssinian", "Bengal", "Birman", "Bombay", "British Shorthair",
    "Beagle", "Boxer", "Chihuahua", "Pomeranian", "Pug",
]
assert set(SELECTED_CLASSES).issubset(set(trainval_raw.classes)), "Nama kelas tidak ditemukan di dataset"

base_dataset = ConcatDataset([trainval_raw, test_raw])
all_original_labels = np.asarray(trainval_raw._labels + test_raw._labels, dtype=np.int64)
assert len(base_dataset) == len(all_original_labels)

original_ids = [trainval_raw.class_to_idx[name] for name in SELECTED_CLASSES]
old_to_new = {original: new for new, original in enumerate(original_ids)}
selected_indices = np.flatnonzero(np.isin(all_original_labels, original_ids))
selected_labels = np.asarray([old_to_new[int(all_original_labels[i])] for i in selected_indices])

train_idx, remainder_idx, train_y, remainder_y = train_test_split(
    selected_indices, selected_labels,
    test_size=0.30, stratify=selected_labels, random_state=SEED,
)
val_idx, test_idx, val_y, test_y = train_test_split(
    remainder_idx, remainder_y,
    test_size=0.50, stratify=remainder_y, random_state=SEED,
)

assert not (set(train_idx) & set(val_idx) or set(train_idx) & set(test_idx) or set(val_idx) & set(test_idx))
assert len(train_idx) + len(val_idx) + len(test_idx) == len(selected_indices)
assert all(len(np.unique(y)) == 10 for y in (train_y, val_y, test_y))

split_rows = []
for split_name, indices in (("train", train_idx), ("validation", val_idx), ("test", test_idx)):
    for idx in indices:
        label = old_to_new[int(all_original_labels[idx])]
        split_rows.append({
            "global_index": int(idx), "split": split_name,
            "label": label, "class_name": SELECTED_CLASSES[label],
        })
pd.DataFrame(split_rows).to_csv(RESULTS_DIR / "dataset_split_seed42.csv", index=False)

configuration = {
    "seed": SEED, "selected_classes": SELECTED_CLASSES, "split": [0.70, 0.15, 0.15],
    "batch_size": BATCH_SIZE, "max_epochs": MAX_EPOCHS, "patience": PATIENCE,
    "weight_decay": WEIGHT_DECAY, "optimizer": "AdamW", "criterion": "CrossEntropyLoss",
    "architecture": "ResNet18", "pretrained": "IMAGENET1K_V1 for B2/B3/B4",
    "image_size": [224, 224], "train_count": len(train_idx),
    "validation_count": len(val_idx), "test_count": len(test_idx),
}
(RESULTS_DIR / "experiment_config.json").write_text(json.dumps(configuration, indent=2), encoding="utf-8")
```

```python
MEAN = WEIGHTS.transforms().mean
STD = WEIGHTS.transforms().std

train_aug_transforms = transforms.Compose([
    transforms.Resize(256),
    transforms.RandomResizedCrop(224, scale=(0.8, 1.0)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor(),
    transforms.Normalize(mean=MEAN, std=STD),
])
eval_transforms = WEIGHTS.transforms()


class PetSubset(Dataset):
    def __init__(self, base, indices, lookup, transform):
        self.base = base
        self.indices = list(map(int, indices))
        self.lookup = lookup
        self.transform = transform

    def __len__(self):
        return len(self.indices)

    def __getitem__(self, position):
        image, old_label = self.base[self.indices[position]]
        return self.transform(image), self.lookup[int(old_label)]


def make_loaders(augmentation):
    train_tf = train_aug_transforms if augmentation else eval_transforms
    options = {"batch_size": BATCH_SIZE, "num_workers": NUM_WORKERS,
               "pin_memory": DEVICE.type == "cuda"}
    generator = torch.Generator().manual_seed(SEED)
    return (
        DataLoader(PetSubset(base_dataset, train_idx, old_to_new, train_tf),
                   shuffle=True, generator=generator, **options),
        DataLoader(PetSubset(base_dataset, val_idx, old_to_new, eval_transforms),
                   shuffle=False, **options),
        DataLoader(PetSubset(base_dataset, test_idx, old_to_new, eval_transforms),
                   shuffle=False, **options),
    )


def make_clean_train_loader():
    """Untuk menghitung generalization gap B4 dengan transformasi train yang identik (tanpa augmentasi)."""
    return DataLoader(
        PetSubset(base_dataset, train_idx, old_to_new, eval_transforms),
        batch_size=BATCH_SIZE, shuffle=False, num_workers=NUM_WORKERS,
        pin_memory=DEVICE.type == "cuda",
    )
```

```python
def make_resnet(condition):
    if condition not in ("B1", "B2", "B3"):
        raise ValueError(f"Kondisi tidak dikenal: {condition}")
    set_seed()
    weights = None if condition == "B1" else WEIGHTS
    model = models.resnet18(weights=weights)

    if condition != "B1":
        for p in model.parameters():
            p.requires_grad = False

    n_features = model.fc.in_features
    model.fc = nn.Linear(n_features, 10)  # head baru secara default trainable

    if condition == "B3":
        for p in model.layer4.parameters():
            p.requires_grad = True

    if condition == "B3":
        param_groups = [
            {"params": model.layer4.parameters(), "lr": 1e-4},
            {"params": model.fc.parameters(), "lr": 1e-3},
        ]
        lr_info = "layer4: 1e-4; fc: 1e-3"
    elif condition == "B2":
        param_groups = [{"params": model.fc.parameters(), "lr": 1e-3}]
        lr_info = "fc: 1e-3"
    else:
        param_groups = [{"params": model.parameters(), "lr": 1e-3}]
        lr_info = "semua layer: 1e-3"

    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total = sum(p.numel() for p in model.parameters())
    if condition == "B2":
        assert trainable == 5130, f"B2 harus 5.130 parameter, tetapi ditemukan {trainable}"
    return model, param_groups, lr_info, trainable, total


def set_train_mode_resnet(model, condition):
    if condition == "B1":
        model.train()  # Semua layer dan BatchNorm belajar
    elif condition == "B2":
        model.eval()   # Backbone frozen: BatchNorm statistics tidak berubah
        model.fc.train()
    elif condition == "B3":
        model.eval()   # Backbone bagian awal frozen
        model.layer4.train()
        model.fc.train()
```

```python
def run_epoch(model, loader, criterion, optimizer=None, train_mode_fn=None, collect_confidence=False):
    is_train = optimizer is not None
    if is_train:
        assert train_mode_fn is not None
        train_mode_fn(model)
    else:
        model.eval()

    total_loss, total_correct, total_samples = 0.0, 0, 0
    pred_all, true_all, conf_all = [], [], []

    with torch.set_grad_enabled(is_train):
        for x, y in loader:
            x = x.to(DEVICE, non_blocking=True)
            y = y.to(DEVICE, non_blocking=True)
            if is_train:
                optimizer.zero_grad(set_to_none=True)
            logits = model(x)
            loss = criterion(logits, y)
            if is_train:
                loss.backward()
                optimizer.step()

            preds = logits.argmax(dim=1)
            batch_size = y.size(0)
            total_loss += float(loss.item()) * batch_size
            total_correct += int((preds == y).sum().item())
            total_samples += batch_size
            if not is_train:
                pred_all.extend(preds.cpu().tolist())
                true_all.extend(y.cpu().tolist())
                if collect_confidence:
                    conf_all.extend(logits.softmax(dim=1).max(dim=1).values.cpu().tolist())

    metrics = {"loss": total_loss / total_samples, "accuracy": total_correct / total_samples}
    if not is_train:
        metrics["macro_f1"] = f1_score(true_all, pred_all, labels=list(range(10)), average="macro", zero_division=0)
        metrics["y_true"] = true_all
        metrics["y_pred"] = pred_all
        if collect_confidence:
            metrics["confidence"] = conf_all
    return metrics


def train_resnet(experiment_key, condition, augmentation=True):
    set_seed()
    loader_train, loader_val, _ = make_loaders(augmentation=augmentation)
    model, groups, lr_info, trainable, total = make_resnet(condition)
    model = model.to(DEVICE)
    optimizer = torch.optim.AdamW(groups, weight_decay=WEIGHT_DECAY)
    criterion = nn.CrossEntropyLoss()

    best_loss = float("inf")
    best_epoch = 0
    waiting = 0
    history = []
    ckpt_path = RESULTS_DIR / f"{experiment_key}_best.pt"
    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    t0 = time.perf_counter()

    for epoch in range(1, MAX_EPOCHS + 1):
        train_metrics = run_epoch(
            model, loader_train, criterion, optimizer=optimizer,
            train_mode_fn=lambda m: set_train_mode_resnet(m, condition),
        )
        val_metrics = run_epoch(model, loader_val, criterion)
        history.append({
            "epoch": epoch,
            "train_loss": train_metrics["loss"],
            "train_accuracy": train_metrics["accuracy"],
            "val_loss": val_metrics["loss"],
            "val_accuracy": val_metrics["accuracy"],
        })

        if val_metrics["loss"] < best_loss - 1e-4:
            best_loss = val_metrics["loss"]
            best_epoch = epoch
            waiting = 0
            torch.save(model.state_dict(), ckpt_path)
        else:
            waiting += 1
            if waiting >= PATIENCE:
                break

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    duration_sec = time.perf_counter() - t0
    history_df = pd.DataFrame(history)
    history_df.to_csv(RESULTS_DIR / f"{experiment_key}_history.csv", index=False)
    selected = history_df.loc[history_df["epoch"] == best_epoch].iloc[0]
    summary = {
        "eksperimen": experiment_key,
        "konfigurasi": condition,
        "augmentasi": augmentation,
        "trainable_parameters": trainable,
        "total_parameters": total,
        "learning_rate": lr_info,
        "epoch_berhenti": len(history),
        "epoch_terbaik": best_epoch,
        "durasi_detik": round(duration_sec, 2),
        "validation_accuracy": float(selected["val_accuracy"]),
        "validation_loss": float(selected["val_loss"]),
        "validation_accuracy_akhir": float(history_df.iloc[-1]["val_accuracy"]),
    }
    del model, optimizer
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()
    return {"summary": summary, "history": history_df, "checkpoint": ckpt_path, "condition": condition}


def load_resnet_checkpoint(run):
    model, _, _, _, _ = make_resnet(run["condition"])
    state = torch.load(run["checkpoint"], map_location="cpu", weights_only=True)
    model.load_state_dict(state)
    return model.to(DEVICE)


def evaluate_resnet_test(run):
    _, _, loader_test = make_loaders(augmentation=False)
    model = load_resnet_checkpoint(run)
    result = run_epoch(model, loader_test, nn.CrossEntropyLoss(), collect_confidence=True)
    del model
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()
    return result
```

```python
runs = {}
for condition in ("B1", "B2", "B3"):
    runs[condition] = train_resnet(condition, condition, augmentation=True)
```

```python
chosen_condition = max(
    ("B2", "B3"),
    key=lambda key: (runs[key]["summary"]["validation_accuracy"],
                     -runs[key]["summary"]["validation_loss"]),
)
runs["B4_no_aug"] = train_resnet("B4_no_aug", chosen_condition, augmentation=False)
runs["B4_with_aug"] = train_resnet("B4_with_aug", chosen_condition, augmentation=True)
```

```python
experiment_keys = ("B1", "B2", "B3", "B4_no_aug", "B4_with_aug")

# Tetapkan model akhir berdasarkan validation sebelum membaca hasil test.
best_key = max(
    experiment_keys,
    key=lambda key: (runs[key]["summary"]["validation_accuracy"],
                     -runs[key]["summary"]["validation_loss"]),
)

test_metrics = {key: evaluate_resnet_test(runs[key]) for key in experiment_keys}
rows = []
for key in experiment_keys:
    record = runs[key]["summary"].copy()
    record["test_accuracy"] = test_metrics[key]["accuracy"]
    record["test_macro_f1"] = test_metrics[key]["macro_f1"]
    rows.append(record)

results_table = pd.DataFrame(rows)
results_table.to_csv(RESULTS_DIR / "experiment_metrics.csv", index=False)

display(results_table[[
    "eksperimen", "konfigurasi", "augmentasi", "trainable_parameters",
    "learning_rate", "epoch_berhenti", "epoch_terbaik", "durasi_detik",
    "validation_accuracy", "test_macro_f1",
]])
```

```python
best_test = test_metrics[best_key]
incorrect_positions = [
    i for i, (actual, predicted) in enumerate(zip(best_test["y_true"], best_test["y_pred"]))
    if actual != predicted
]

n_examples = min(8, len(incorrect_positions))
sample_positions = sorted(random.Random(SEED).sample(incorrect_positions, n_examples))
failure_rows = []

for position in sample_positions:
    actual = int(best_test["y_true"][position])
    predicted = int(best_test["y_pred"][position])
    confidence = float(best_test["confidence"][position])
    failure_rows.append({
        "model": best_key,
        "global_index": int(test_idx[position]),
        "ground_truth": SELECTED_CLASSES[actual],
        "prediksi": SELECTED_CLASSES[predicted],
        "confidence": confidence,
        "hipotesis_penyebab": "",
    })

pd.DataFrame(failure_rows, columns=[
    "model", "global_index", "ground_truth", "prediksi", "confidence", "hipotesis_penyebab",
]).to_csv(RESULTS_DIR / "failure_analysis.csv", index=False)

if sample_positions:
    fig, axes = plt.subplots(2, 4, figsize=(16, 8.5))
    for ax in axes.flat:
        ax.axis("off")
    for ax, position in zip(axes.flat, sample_positions):
        image, _ = base_dataset[int(test_idx[position])]
        actual = int(best_test["y_true"][position])
        predicted = int(best_test["y_pred"][position])
        ax.imshow(image)
        ax.set_title(
            f"Aktual: {SELECTED_CLASSES[actual]}\n"
            f"Prediksi: {SELECTED_CLASSES[predicted]}\n"
            f"Confidence: {best_test['confidence'][position]:.1%}",
            fontsize=10,
        )
    fig.suptitle(f"Failure Analysis — Model terbaik: {best_key}", fontsize=14)
    fig.tight_layout()
    fig.savefig(RESULTS_DIR / "failure_analysis.png", dpi=160, bbox_inches="tight")
    plt.show()
    plt.close(fig)
```

```python
RUN_B5_BONUS = False


def make_efficientnet(strategy):
    set_seed()
    weights = None if strategy == "B1" else models.EfficientNet_B0_Weights.IMAGENET1K_V1
    model = models.efficientnet_b0(weights=weights)
    if strategy in ("B2", "B3"):
        for parameter in model.parameters():
            parameter.requires_grad = False

    features = model.classifier[-1].in_features
    model.classifier[-1] = nn.Linear(features, 10)
    if strategy == "B1":
        groups = [{"params": model.parameters(), "lr": 1e-3}]
        lr_text = "semua layer: 1e-3"
    elif strategy == "B2":
        groups = [{"params": model.classifier.parameters(), "lr": 1e-3}]
        lr_text = "classifier: 1e-3"
    else:
        for parameter in model.features[-2:].parameters():
            parameter.requires_grad = True
        groups = [
            {"params": model.features[-2:].parameters(), "lr": 1e-4},
            {"params": model.classifier.parameters(), "lr": 1e-3},
        ]
        lr_text = "blok fitur terakhir: 1e-4; classifier: 1e-3"
    return model, groups, lr_text


def train_efficientnet_bonus():
    strategy = runs[best_key]["condition"]
    augmentation = runs[best_key]["summary"]["augmentasi"]
    train_loader, val_loader, test_loader = make_loaders(augmentation)
    model, groups, lr_info = make_efficientnet(strategy)
    model = model.to(DEVICE)
    trainable = sum(p.numel() for p in model.parameters() if p.requires_grad)
    total = sum(p.numel() for p in model.parameters())
    optimizer = torch.optim.AdamW(groups, weight_decay=WEIGHT_DECAY)
    criterion = nn.CrossEntropyLoss()
    best_loss, best_epoch, patience_count = float("inf"), 0, 0
    history = []
    ckpt = RESULTS_DIR / "B5_EfficientNetB0_best.pt"

    def mode_train(m):
        if strategy == "B1":
            m.train()
        else:
            m.eval()
            m.classifier.train()
            if strategy == "B3":
                m.features[-2:].train()

    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    started = time.perf_counter()
    for epoch in range(1, MAX_EPOCHS + 1):
        train_result = run_epoch(model, train_loader, criterion, optimizer, mode_train)
        val_result = run_epoch(model, val_loader, criterion)
        history.append({
            "epoch": epoch, "train_loss": train_result["loss"],
            "train_accuracy": train_result["accuracy"],
            "val_loss": val_result["loss"], "val_accuracy": val_result["accuracy"],
        })
        if val_result["loss"] < best_loss - 1e-4:
            best_loss, best_epoch, patience_count = val_result["loss"], epoch, 0
            torch.save(model.state_dict(), ckpt)
        else:
            patience_count += 1
            if patience_count >= PATIENCE:
                break
    if DEVICE.type == "cuda":
        torch.cuda.synchronize()
    elapsed = time.perf_counter() - started

    model.load_state_dict(torch.load(ckpt, map_location="cpu", weights_only=True))
    test_result = run_epoch(model, test_loader, criterion)
    del model, optimizer
    if DEVICE.type == "cuda":
        torch.cuda.empty_cache()

    hist_df = pd.DataFrame(history)
    hist_df.to_csv(RESULTS_DIR / "B5_EfficientNetB0_history.csv", index=False)
    chosen = hist_df.loc[hist_df["epoch"] == best_epoch].iloc[0]
    return {
        "eksperimen": "B5_EfficientNetB0", "konfigurasi": f"EfficientNet-B0/{strategy}",
        "augmentasi": augmentation, "trainable_parameters": trainable,
        "total_parameters": total, "learning_rate": lr_info,
        "epoch_berhenti": len(hist_df), "epoch_terbaik": best_epoch,
        "durasi_detik": round(elapsed, 2),
        "validation_accuracy": float(chosen["val_accuracy"]),
        "validation_loss": float(chosen["val_loss"]),
        "test_macro_f1": test_result["macro_f1"],
        "test_accuracy": test_result["accuracy"],
    }


if RUN_B5_BONUS:
    bonus_row = train_efficientnet_bonus()
    bonus_table = pd.concat([results_table, pd.DataFrame([bonus_row])], ignore_index=True)
    bonus_table.to_csv(RESULTS_DIR / "experiment_metrics_with_bonus.csv", index=False)
    display(bonus_table[[
        "eksperimen", "konfigurasi", "total_parameters", "trainable_parameters",
        "learning_rate", "epoch_berhenti", "durasi_detik",
        "validation_accuracy", "test_macro_f1",
    ]])
```
